# Flash Drought Indonesia — Revised Workflow and Audit

**Purpose.** This notebook is the scientific control document for the 1995–2025 Indonesia flash-drought workflow.  
It preserves useful outputs from notebooks 01–04, but changes which products are treated as the **benchmark** and which are treated as **sensitivity / diagnostics**.

## Main scientific decision

For the current thesis stage, the benchmark is a **soil-moisture rapid-intensification definition**, not a new composite index.

Primary benchmark:
- pentad RZSM (0–100 cm);
- start from ≥ P40;
- reach < P20 rapidly;
- mean decline rate ≥ 5 percentile points per pentad;
- require persistence of drought conditions rather than accepting every brief P20 crossing;
- meteorological variables (precipitation, T2m, VPD) are **drivers/diagnostics**, not part of the primary detection rule.

The weighted multivariate index remains a later-stage development.


## What is retained from the existing notebooks

### Notebook 01 — retain
- ERA5-Land hourly SWVL1–SWVL3 → strict daily means.
- RZSM 0–100 cm = `0.07*SWVL1 + 0.21*SWVL2 + 0.72*SWVL3`.
- no-leap calendar.
- 73 non-overlapping 5-day pentads.
- calendar-pentad soil-moisture percentile.

**Revision:** the exact-pentad percentile is retained as the baseline, but percentile construction must be included in sensitivity testing because 1995–2025 provides only 31 values per calendar pentad.

### Notebook 02 — retain as broad screening / event-engine source
The event bookkeeping, phase flags, onset speed, duration, severity, crossing/minimum/termination indices are useful.

**Do not use the old broad M0 inventory as the paper benchmark.**

### Notebook 03b — reuse the expensive detections
The existing `M2A` output is the closest already-computed inventory to the intended soil-moisture benchmark because it combines rapid depletion with ≥3 initial pentads below P20.

For paper-ready results this pack therefore aliases:
`M2A -> FD_BENCHMARK`

`M0/M1/M2B/M3` remain sensitivity / segmentation experiments.

### Notebook 03c — keep as secondary regional analysis
Macro-islands are useful for communicating regional contrasts, but they are geographic aggregates, not homogeneous climate regimes. Grid-cell climatology remains the primary result.

### Notebook 04 — replace the inferential logic
The old all-event composite pools hundreds of thousands of adjacent grid-cell starts. That is useful descriptively but must not be treated as independent samples.

The revised driver notebook:
1. keeps the grid-level meteorological anomaly cube already produced by Notebook 04;
2. uses the benchmark detection instead of M1;
3. computes one composite value per **year** first, then averages years equally;
4. bootstraps across years;
5. separates preconditioning (`lag -3:-1`) from rapid-development (`lag 0...`) rather than calling lag 0 “pre-onset”;
6. adds phase-based standardized P/T/VPD diagnostics.


## Final workflow

```text
ERA5-Land 1995–2025
  ├─ SWVL1, SWVL2, SWVL3
  ├─ precipitation
  ├─ T2m
  └─ D2m
        ↓
strict daily preprocessing
        ↓
73 fixed pentads yr⁻¹
        ↓
RZSM 0–100 cm             T2m + D2m
        ↓                      ↓
seasonal SM percentile        VPD
        ↓
rapid RZSM depletion + drought persistence
        ↓
FD_BENCHMARK
        ├─ climatological occurrence / frequency
        ├─ onset seasonality
        ├─ onset speed
        ├─ total duration + P20 persistence
        ├─ severity
        ├─ hotspot
        └─ annual spatial metrics / trend
        ↓
meteorological driver diagnosis
        ├─ precipitation anomaly
        ├─ T anomaly
        └─ VPD anomaly
        ↓
year-blocked composites + uncertainty
        ↓
sensitivity / consistency
        ↓
Indonesia flash-drought framework

Composite multivariate index = later stage
```


## Output definitions that should appear in the thesis/paper

1. **Occurrence probability**: percentage of complete years in which a grid cell experiences ≥1 benchmark FD onset.
2. **Frequency**: mean benchmark events per decade at each grid cell.
3. **Seasonality**: area-weighted monthly/pentad contribution of FD starts.
4. **Onset speed**: percentile-point decline per pentad.
5. **Duration**: report both total event duration and initial P20 persistence to remove ambiguity.
6. **Severity**: cumulative RZSM percentile deficit below P40; P20-deficit is retained as a secondary severity measure.
7. **Hotspot**: high-occurrence grid cells (quantile-based); report sensitivity to the hotspot percentile instead of inventing a weighted index.
8. **Annual metrics**: annual onset exposure, maximum onset area, and maximum drought-core area.
9. **Drivers**: anomalies during preconditioning and rapid-development phases with year-blocked uncertainty.

Do not refer to the current sensitivity comparison as “validation”; use **sensitivity**, **consistency**, or **robustness**.
